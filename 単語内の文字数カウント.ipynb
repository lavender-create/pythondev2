```python

# 空のリストを作成（英単語を格納）
sample = []

# 未入力になるまで英単語を繰り返し受け付け
while True:
    i= input("英単語を入力してください：")
    if  i == "":
        break  # 未入力でEnter → 終了

    sample.append(i)  # 単語をリストに追加
    print(f"英単語を入力してください：{i}")

# 結果を表示
sample.sort()
print(f"入力した英単語：{sample}")

alfa_code = dict.fromkeys(list("abcdefghijklmnopqrstuvwxyz"), 0)
animal=[]
for i in sample:
    for j in i:
        alfa_code[j]+=1
    animal.append(j)
    # items

for key, value in alfa_code.items():
    if value >0:
        print(f"{key} が {value}個ありました")
```
