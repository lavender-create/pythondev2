```python
class Baseball_Team :
#勝率を計算するメソッド
    def  calc_win_rate(self):
        return self.win /(self.win+self.lose)

    def __init__(self, name, win,lose, draw):
        # property
        self.name = name
        self.win = win
        self.lose = lose
        self.draw = draw

    def show_team_result(self):
        win_rate = self.calc_win_rate()
        print(f"{self.name}： win {self.win}、lose {self.lose}、draw {self.draw}、rate {win_rate:.3f}")

# チームデータ（名前, 勝ち, 負け, 引き分け）
teams = [
    ("Archers", 77, 64, 2),
    ("Meteors", 71, 69, 3),
    ("Cheetahs", 69, 68, 6),
    ("Sharks", 70, 70, 3),
    ("Pythons", 68, 73, 2),
    ("Owls", 59, 82, 2)
]
# インスタンス生成
team_list=[]
for name, win, lose, draw in teams:
    team = Baseball_Team(name, win, lose, draw)
    team_list.append(team)

    # 各チームの結果を表示
for team in team_list:
    team.show_team_result()
```
