```python
# 点数の合計値を格納する変数
total_score = 0
# 人数を数えるカウンタ変数
count = 0

while True:
    score = int(input("点数を入力してください："))
    
    if score == -1:   # -1 が入力されたら終了
        break
    
    # 合計に加算
    total_score += score
    # 人数カウンタを1増やす
    count += 1
    print(f"点数を入力してください：{int(score)}")
# 人数が0でなければ平均を計算
if count > 0:
    average = total_score / count
    # 小数第2位を四捨五入 → 小数第1位まで表示
    average = round(average, 1)
    print(f"{count}人のテストの平均点は{average}点です")
else:
    print("有効な点数が入力されませんでした。")
```
